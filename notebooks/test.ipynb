```python
import tensorflow as tf
from openvino.runtime import Core

# Check for GPU availability
physical_devices = tf.config.list_physical_devices('GPU')
print(f"Available GPUs: {physical_devices}")

# Create a simple model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(128, activation='relu', input_shape=(10,)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Compile model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Generate dummy training data
import numpy as np
X_train = np.random.rand(1000, 10)
y_train = np.random.randint(2, size=(1000, 1))

# Train model
model.fit(X_train, y_train, epochs=5, batch_size=32)

# Save model
model.save('model.keras')

# Convert model to OpenVINO format
from openvino.tools.ovc import convert_model
from openvino.runtime import Core


print("Model successfully compiled and loaded on Intel UHD Graphics!")
```

```python

# Convert TensorFlow model to OpenVINO format
ov_model = convert_model("model.keras")  # Remove framework argument

# Save the converted model
ov_model.serialize("model.xml", "model.bin")

# Load and compile the model on Intel UHD Graphics
ie = Core()
compiled_model = ie.compile_model("model.xml", "GPU")
```
